# EDA - IWSLT'15 English-Vietnamese corpus

Real parallel corpus of TED-talk transcripts (Luong & Manning, IWSLT 2015): 133,317 train / 1,268 validation / 1,268 test sentence pairs, mirrored from HuggingFace since the original Stanford NLP host blocks automated downloads. See `src/data/download.py`.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt

from src import config
from src.data.vocab import tokenize

train_df = pd.read_parquet(config.TRAIN_PARQUET)
val_df = pd.read_parquet(config.VAL_PARQUET)
test_df = pd.read_parquet(config.TEST_PARQUET)
print('train:', len(train_df), 'val:', len(val_df), 'test:', len(test_df))
train_df.head()
```

```python
en_lens = train_df['en'].apply(lambda s: len(tokenize(s)))
vi_lens = train_df['vi'].apply(lambda s: len(tokenize(s)))

print(en_lens.describe())
print(vi_lens.describe())
print('pct of pairs with both sides <= 30 tokens (the MAX_LEN used for training):',
      ((en_lens <= 30) & (vi_lens <= 30)).mean())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(en_lens, bins=50)
axes[0].set_title('English sentence length (tokens)')
axes[1].hist(vi_lens, bins=50, color='orange')
axes[1].set_title('Vietnamese sentence length (tokens)')
for ax in axes:
    ax.set_xlabel('tokens')
    ax.set_ylabel('count')
fig.tight_layout()
fig.savefig(config.REPORTS_DIR / 'figures' / 'sentence_length_distribution.png', dpi=150)
plt.show()
```
